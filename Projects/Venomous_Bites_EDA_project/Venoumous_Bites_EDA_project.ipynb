```python
#percentage of missing values
missing_percent = df.isnull().mean() * 100
print(missing_percent.sort_values(ascending=False))
```

```python
# missing values plot

missing_percent = missing_percent[
    missing_percent > 0
].sort_values(ascending=False)

if len(missing_percent) > 0:
    missing_percent.plot(
        kind="bar",
        figsize=(12, 8)
    )
    
    plt.title("Percentage of Missing Values by Column")
    plt.ylabel("Missing values (%)")
    plt.xlabel("Columns")
    plt.xticks(rotation=45, ha="right")
    plt.tight_layout()
    plt.show()
else:
    print("No missing values.")
```

```python
df['Local_Swelling'] = df['Local_Swelling'].fillna(df['Local_Swelling'].mode()[0])
```

```python
df.isna().sum()
```

```python
num_col = df.select_dtypes(include=(int,float)).columns.to_list()
num_col
```

```python
# correlation of each column
corr_matrix = df[num_col].corr()
corr_matrix
```

```python
# correlation heatmap of each column
sns.heatmap(corr_matrix,annot=True)
plt.title("Heat Map (Correlation of each)")
```

```python
# outliars with boxplot

for c in num_col:
  plt.figure(figsize=(10,8))
  sns.boxplot(x=df[c],color='red')
  plt.title(c)
  plt.show()
```

```python



for col in num_col:
    Q1 = df[col].quantile(0.25)
    Q3 = df[col].quantile(0.75)
    IQR = Q3 - Q1
    lower_bound = Q1 - 1.5 * IQR
    upper_bound = Q3 + 1.5 * IQR
    df[col] = df[col].clip(lower_bound, upper_bound)
```

```python
# Recheck outliars after removal with boxplot

for c in num_col:
  plt.figure(figsize=(10,8))
  sns.boxplot(x=df[c],color='red')
  plt.title(c)
  plt.show()
```

```python
# Number of Records by Bite Source
df["Bite_Source_Target"].value_counts().plot(kind="bar", figsize=(10, 5), color="steelblue")
plt.title("Number of Records by Bite Source")
plt.xlabel("Bite source")
plt.ylabel("Number of records")
plt.tight_layout()
plt.show()
```

```python
# Bite Source Distribution

df["Bite_Source_Target"].value_counts().plot(
    kind="pie",
    autopct="%1.1f%%",
    figsize=(8, 8)
)
plt.ylabel("")
plt.title("Bite Source Distribution")
plt.show()
```

```python
# Average Blood Pressure by Bite Source
df.groupby("Bite_Source_Target")["Blood_Pressure_Systolic"].mean().plot(
    kind="bar", figsize=(10, 5), color="red"
)

plt.title("Average Blood Pressure by Bite Source")
plt.xlabel("Bite source")
plt.ylabel("Average systolic blood pressure")
plt.xticks(rotation=45, ha="right")
plt.tight_layout()
plt.show()
```

```python
# Bite_Source_Target vs Local_Swelling
plt.scatter(df["Bite_Source_Target"], df["Local_Swelling"], alpha=0.3,)
swelling_colors = df["Local_Swelling"].map({
    "Mild": "green",
    "Moderate": "orange",
    "Severe": "red"
}).fillna("gray")

scatter = plt.gca().collections[-1]
scatter.set_color(swelling_colors)
scatter.set_alpha(0.3)

plt.xlabel("Bite_Source_Target")
plt.ylabel("Local_Swelling")
plt.title("Bite_Source_Target vs Local_Swelling")
plt.show()
```
