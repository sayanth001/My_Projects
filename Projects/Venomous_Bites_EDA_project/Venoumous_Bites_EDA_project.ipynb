```python
num_col = df.select_dtypes(include=(int,float)).columns.to_list()
num_col
```

```python
# correlation of each column
corr_matrix = df[num_col].corr()
corr_matrix
```

```python
# correlation of each column - heatmap
sns.heatmap(corr_matrix,annot=True)
plt.title("Heat Map (Correlation of each)")
```

```python
# outliars with boxplot

for c in num_col:
  plt.figure(figsize=(10,8))
  sns.boxplot(x=df[c],color='red')
  plt.title(c)
  plt.show()
```

```python



for col in num_col:
    Q1 = df[col].quantile(0.25)
    Q3 = df[col].quantile(0.75)
    IQR = Q3 - Q1
    lower_bound = Q1 - 1.5 * IQR
    upper_bound = Q3 + 1.5 * IQR
    df[col] = df[col].clip(lower_bound, upper_bound)
```

```python
# Recheck outliars after removal with boxplot

for c in num_col:
  plt.figure(figsize=(10,8))
  sns.boxplot(x=df[c],color='red')
  plt.title(c)
  plt.show()
```

```python
# Number of Records by Bite Source - barchart

plt.figure(figsize=(10, 5))

sns.countplot(data=df, x='Bite_Source_Target',hue='Bite_Source_Target')

plt.title('Number of Records by Bite Source', fontsize=14)
plt.xlabel('Bite Source', fontsize=12)
plt.ylabel('Number of Patients', fontsize=12)

plt.show()
```

```python
# Percentage of Patients by Bite Source - piechart

bite_counts = df['Bite_Source_Target'].value_counts()
plt.figure(figsize=(8, 8))
plt.pie(
    bite_counts,
    labels=bite_counts.index,
    autopct='%1.1f%%',
    startangle=140,
    colors=['red','green','yellow','blue']
)
plt.title('Percentage of Patients by Bite Source', fontsize=14, fontweight='bold')
plt.show()
```

```python
# Time Since Bite vs. Heart Rate (Sample of 500 Patients) - scatterplot
plt.figure(figsize=(10, 6))

df_sample = df.sample(n=500, random_state=42)

sns.scatterplot(data=df_sample, x='Time_Since_Bite_Min', y='Heart_Rate_BPM', hue='Bite_Source_Target')
plt.title('Time Since Bite vs. Heart Rate (Sample of 500 Patients)', fontsize=14)
plt.xlabel('Time Since Bite (Minutes)', fontsize=12)
plt.ylabel('Heart Rate (BPM)', fontsize=12)
plt.legend(title='Bite Source')
plt.grid()
plt.show()
```

```python
# Count of Bites by Bite Source and Gender - countplot

plt.figure(figsize=(10, 6))
sns.countplot(data=df, x='Bite_Source_Target', hue='Gender')
plt.title('Count of Bites by Bite Source and Gender')
plt.xlabel('Bite Source')
plt.ylabel('Number of Patients')
plt.legend(title='Gender')
plt.xticks(rotation=15)
plt.tight_layout()
plt.show()
```

```python
# Average Heart Rate by Age - lineplot

plt.figure(figsize=(12, 6))
age_heart_rate = df.groupby('Age')['Heart_Rate_BPM'].mean().reset_index()
sns.lineplot(data=age_heart_rate, x='Age', y='Heart_Rate_BPM', color='green')
plt.title('Average Heart Rate by Age', fontsize=14)
plt.xlabel('Age of Patient', fontsize=12)
plt.ylabel('Average Heart Rate (BPM)', fontsize=12)
plt.grid()
plt.show()
```
